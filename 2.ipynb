```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
x = [1, 2, 3, 4, 5]
y = [2, 4, 6, 8, 10]

plt.plot(x, y)
plt.xlabel("X Axis")
plt.ylabel("Y Axis")
plt.title("Simple Line Plot")
plt.show()
```

```python
x = np.linspace(0, 10, 20)

plt.figure(figsize=(12, 6))

plt.plot(x, x, 'b-', label="Solid line", linewidth=2)
plt.plot(x, x+1, 'r--', label="Dashed line", linewidth=2)
plt.plot(x, x+2, 'g-', label="Dash-dot line", linewidth=2)
plt.plot(x, x+3, 'm', label="Dotted line", linewidth=2)
plt.plot(x, x+4, 'co-', label="Circles", markersize=8)
plt.plot(x, x+5, 's-', label="Squares", markersize=6)
plt.plot(x, x+6, '^-', label="Triangles", markersize=6)

plt.xlabel("X Axis")
plt.ylabel("Y Axis")
plt.title("Line Styles and Markers")
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

```python
x = list(range(-4, 5))
y = x

plt.figure(figsize=(6, 4))
plt.plot(x, y, marker="o", color="red")

plt.axhline(0, color="gray", linewidth=0.8)
plt.axvline(0, color="gray", linewidth=0.8)

plt.title("Graph y=x")
plt.xlabel("X")
plt.ylabel("Y")

plt.show()
```

```python
x = list(range(-4, 5))
y = [i * i for i in x]

plt.figure(figsize=(6, 4))
plt.plot(x, y, marker="o", color="red")

plt.axhline(0, color="gray", linewidth=0.8)
plt.axvline(0, color="gray", linewidth=0.8)

plt.title("Parabola Graph")
plt.xlabel("X")
plt.ylabel("Y")

plt.show()
```

```python
x = list(range(-4, 5))
y = [i ** 3 for i in x]

plt.figure(figsize=(6, 4))
plt.plot(x, y, marker="o", color="red")

plt.axhline(0, color="gray", linewidth=0.8)
plt.axvline(0, color="gray", linewidth=0.8)

plt.title("Cubic Graph")
plt.xlabel("X")
plt.ylabel("Y")

plt.show()
```

```python
x = list(range(50))
y = [i ** 0.5 for i in x]

plt.figure(figsize=(6, 4))
plt.plot(x, y, marker="o", color="red")

plt.axhline(0, color="gray", linewidth=0.8)
plt.axvline(0, color="gray", linewidth=0.8)

plt.title("Square Root Graph")
plt.xlabel("X")
plt.ylabel("Y")

plt.show()
```

```python
x = list(range(50))
y = np.sqrt(x)

plt.figure(figsize=(6, 4))
plt.plot(x, y, marker="o", color="red")

plt.axhline(0, color="gray", linewidth=0.8)
plt.axvline(0, color="gray", linewidth=0.8)

plt.title("Square Root Graph")
plt.xlabel("X")
plt.ylabel("Y")

plt.show()
```

```python
x = list(range(-10, 11))
y = np.abs(x)

plt.figure(figsize=(6, 4))
plt.plot(x, y, marker="o", color="red")

plt.axhline(0, color="gray", linewidth=0.8)
plt.axvline(0, color="gray", linewidth=0.8)

plt.title("Absolute Value Graph")
plt.xlabel("X")
plt.ylabel("Y")

plt.show()
```

```python
x = np.linspace(0, 2 * np.pi, 500)
y = np.sin(x)

plt.figure(figsize=(6, 4))
plt.plot(x, y, color="red", linewidth=1)

plt.axhline(0, color="gray", linewidth=0.8)
plt.axvline(0, color="gray", linewidth=0.8)

plt.title("y = sin(x)")
plt.xlabel("x (radians)")
plt.ylabel("sin(x)")
plt.show()
```

```python
x = np.linspace(0, 2 * np.pi, 500)
y = np.cos(x)

plt.figure(figsize=(6, 4))
plt.plot(x, y, color="red", linewidth=1)

plt.axhline(0, color="gray", linewidth=0.8)
plt.axvline(0, color="gray", linewidth=0.8)

plt.title("y = cos(x)")
plt.xlabel("x (radians)")
plt.ylabel("cos(x)")
plt.show()
```

```python
x = np.linspace(0, 2 * np.pi, 500)
y = np.tan(x)

y[np.abs(y) > 10] = np.nan # Removes vertical lines

plt.figure(figsize=(6, 4))
plt.plot(x, y, color="red", linewidth=1)
plt.ylim(-3, 3)

plt.axhline(0, color="gray", linewidth=0.8)
plt.axvline(0, color="gray", linewidth=0.8)

# Dashed lines for asymptotes
for k in [-3, -1, 1, 3]:
    plt.axvline(k * np.pi / 2, color="gray", linestyle="--", alpha=0.5)

plt.title("y = tan(x)")
plt.xlabel("x (radians)")
plt.ylabel("tan(x)")
plt.show()
```

```python
x = np.linspace(0.01, 2 * np.pi - 0.01, 500)
y = 1 / np.tan(x)

y[np.abs(y) > 10] = np.nan # Removes vertical lines

plt.figure(figsize=(6, 4))
plt.plot(x, y, color="red", linewidth=1)
plt.ylim(-6, 6)

plt.axhline(0, color="gray", linewidth=0.8)
plt.axvline(0, color="gray", linewidth=0.8)

# Dashed lines for asymptotes
for k in [0, 1, 2, 3]:
    plt.axvline(k * np.pi / 2, color="gray", linestyle="--", alpha=0.5)

plt.title("y = cot(x)")
plt.xlabel("x (radians)")
plt.ylabel("cot(x)")
plt.show()
```

```python
x = np.linspace(0.01, 10, 500)
y = np.log(x)

plt.figure(figsize=(8, 6))
plt.plot(x, y, color="teal", linewidth=2)
plt.axhline(0, color="gray", linewidth=0.8)
plt.axvline(0, color="gray", linewidth=0.8)

plt.scatter([1], [0], color="red", zorder=5)
plt.annotate("(1, 0)", xy=(1, 0), xytext=(1.5, -1), fontsize=10, color="red")

plt.title("y = ln(x)")
plt.xlabel("X")
plt.ylabel("ln(X)")
plt.grid(True, alpha=0.3)
plt.show()
```

```python
x = np.linspace(-3, 3, 500)
y = np.exp(x)

plt.figure(figsize=(8, 4))
plt.plot(x, y, color="crimson", linewidth=2)
plt.axhline(0, color="gray", linewidth=0.8)
plt.axvline(0, color="gray", linewidth=0.8)

plt.title("y = e^x")
plt.xlabel("x")
plt.ylabel("exp(x)")
plt.grid(True, alpha=0.3)
plt.show()
```

```python
x = np.linspace(-3, 3, 500)

plt.figure(figsize=(9, 5))
plt.plot(x, 2 ** x,  label="2^x",  color="steelblue")
plt.plot(x, np.exp(x), label="e^x ≈ 2.718^x", color="black", linewidth=2)
plt.plot(x, 3 ** x,  label="3^x",  color="darkorange")
plt.plot(x, 10 ** x, label="10^x", color="crimson")

plt.axhline(0, color="gray", linewidth=0.8)
plt.axvline(0, color="gray", linewidth=0.8)
plt.ylim(0, 10)
plt.legend()

plt.title("Comparing exponential functions")
plt.xlabel("x")
plt.ylabel("y")
plt.grid(True, alpha=0.3)
plt.show()
```
