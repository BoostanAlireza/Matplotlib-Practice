```python
import matplotlib.pyplot as plt
```

```python
x = [1, 2, 3, 4, 5, 6]
y = [2, 4, 6, 8, 10, 12]

plt.plot(x, y)
plt.show()
```

```python
plt.plot(x, y, color="green", linestyle="--", marker="o")
plt.title("Simple line plot")
plt.xlabel("X label")
plt.ylabel("Y label")
plt.grid(True)
plt.show()
```

```python
names = ["Ali", "Sara", "Reza", "Neda"]
scores = [88, 92, 79, 95]

plt.bar(names, scores, color="orange")
plt.title("Student score")
plt.xlabel("Name")
plt.ylabel("Scores")
plt.show()
```

```python
ages = [23, 25, 21, 24]
scores = [88, 92, 79, 95]

plt.scatter(ages, scores, color="green", marker="o")
plt.title("Age vs Score (Scatter plot)")
plt.xlabel("Age")
plt.ylabel("Score")
plt.grid(True)
plt.show()
```

```python
names = ["Ali", "Sara", "Reza", "Neda"]
scores = [88, 92, 79, 95]
ages = [23, 25, 21, 24]

plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
plt.bar(names, scores, color="orange")
plt.title("Scores")

plt.subplot(1, 2, 2)
plt.plot(names, ages, marker="o", color="green")
plt.title("Ages")

plt.tight_layout()
plt.show()
```

```python

plt.plot(names, scores, marker="o", label="Score")
plt.plot(names, ages, marker="s", label="Age", color="green")
plt.title("Age and Score comparison")
plt.xlabel("Name")
plt.legend()
plt.show()
```

```python
x = [1, 2, 3, 4, 5]
y = [2, 4, 6, 8, 10]


plt.plot(x, y, color="red", linestyle="--", linewidth=2, marker="o", markersize=8)
plt.title("Customized line chart")
plt.xlabel("X Axis")
plt.ylabel("Y Axis")
plt.grid(True)
plt.show()
```

```python
names = ["Ali", "Sara", "Reza", "Neda"]
scores = [88, 92, 79, 95]
ages = [23, 25, 21, 24]

plt.plot(names, scores, marker="o", label="Score", color="blue")
plt.plot(names, ages, marker="s", label="Age", color="green")
plt.title("Age & Score Comparison")
plt.xlabel("Name")
plt.ylabel("Value")

plt.legend(loc="upper left")

plt.annotate("Highest point", xy=("Neda", 95), xytext=("Reza", 90), arrowprops=dict(facecolor="black", arrowstyle="->"))

plt.grid(True)
plt.show()
```
