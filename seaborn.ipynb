```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
sns.get_dataset_names()
```

```python
mpg = sns.load_dataset("mpg")
```

```python
mpg
```

```python
sns.set_theme()
```

```python
sns.set_style("white") # white, whitegrid, dark, darkgrid, ticks
```

```python
sns.scatterplot(data=mpg, x="weight", y="mpg")
```

```python
mpg.corr(numeric_only=True)
```

```python
sns.heatmap(mpg.corr(numeric_only=True))
```

```python
plt.figure(figsize=(10, 6))
sns.scatterplot(data=mpg, x="weight", y="mpg", hue="origin", palette="mako")
```

```python
sns.lineplot(data=mpg, x="model_year", y="mpg")
```

```python
sns.pointplot(data=mpg, x="model_year", y="mpg")
```

```python
sns.kdeplot(data=mpg, x="weight", y="mpg")
```

```python
sns.kdeplot(data=mpg, x="weight", y="mpg", fill=True)
```

```python
sns.histplot(data=mpg, x="weight", y="mpg")
```

```python
sns.histplot(data=mpg, x="mpg")
```

```python
sns.histplot(mpg["mpg"])
```

```python
sns.histplot(mpg["mpg"], kde=True)
```

```python
sns.histplot(mpg["mpg"], kde=True, bins=30)
```
